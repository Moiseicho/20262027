```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as stats

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

df = pd.read_csv("gfr.txt",header=None)
df.head()

x=df.iloc[:,0].dropna().to_numpy(dtype=float)

mu_hat, std_hat = stats.norm.fit(x)
shape_hat, loc_hat, scale_hat = stats.gamma.fit(x)
shape_st_hat, _, scale_st_hat = stats.gamma.fit(x,floc=0)


log_likelihood_gamma = stats.gamma.logpdf(x,a=shape_hat,loc=loc_hat,scale=scale_hat).sum()
log_likelihood_norm = stats.norm.logpdf(x,mu_hat,std_hat).sum()
log_likelihood_gamma_st = stats.gamma.logpdf(x,a=shape_st_hat,loc=0,scale=scale_st_hat).sum()
xx = np.linspace(0, 110, 500)
gamma_pdf = stats.gamma.pdf(
    xx, a=shape_hat, loc=loc_hat, scale=scale_hat
)
norm_pdf=stats.norm.pdf(
    xx,mu_hat,std_hat
)
gamma_st_pdf = stats.gamma.pdf(
    xx, a=shape_st_hat, loc=0, scale=scale_st_hat
)

df.hist(density=True,range=[0,110], label="data",bins=30);
plt.plot(xx,gamma_pdf, "r-", label="Gamma");
plt.plot(xx,gamma_st_pdf, "r-", color="green", label="Gamma static");
plt.plot(xx,norm_pdf,"r-",color="blue", label="Normal");
plt.legend()
plt.show()

log_likelihood_gamma
log_likelihood_norm
log_likelihood_gamma_st
```
